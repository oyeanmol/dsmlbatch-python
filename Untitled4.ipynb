```python
# Methods

class Student:
    # class variable
    college = "GIT COllege"

    # init constructor with self
    def __init__(self, name, roll_no):
        self.name = name
        self.roll_no = roll_no
        

    # Methods
    def exam(self, name, branch):
        self.name = name
        self.branch = branch
        print("Exam de bhai")
```

```python
s1 = Student("Sadique", 101)
```

```python
s1.exam("Sadique", "CSE")
```

```python
class Human:
    def __init__(self, name):
        self.name = name
    
    def eat(self):
        print(f"{self.name} eats!")

    def listen(self):
        print(f"{self.name} Listens!")
```
