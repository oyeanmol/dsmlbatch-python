## 1. Closures

A **closure** is a nested function that remembers and has access to variables from its enclosing scope, even after the enclosing function has finished execution. In Python, when an inner function is returned by an outer function, the inner function retains access to the variables of the outer function's scope. This is particularly useful for creating functions factory or maintaining state.

```python
# Example of a Closure

def outer_function(x):
  # `x` is a variable in the enclosing scope
  def inner_function(y):
    return x + y
  return inner_function # Return the inner function

# Create a closure
add_five = outer_function(5)

# The 'add_five' function now 'remembers' x=5
result1 = add_five(10)
print(f"add_five(10) = {result1}")

# Create another closure
add_ten = outer_function(10)
result2 = add_ten(10)
print(f"add_ten(10) = {result2}")
```

```python
def greet(name):
  return f"Hello {name}"
```

```python
def decor(func):
  def inner(name):
    print("-" * 10)
    print(func(name))
    print("-" * 10)

  return inner
```

```python
greet("anmol")
```

```python
decor(greet)
```

```python
a = decor(greet)
```

```python
a("Anmol")
```

```python
def div(a,b):
  return a/b
```

```python
def decor1(func):
  def inner(a,b):
    if a < b:
      a,b = b,a
    return func(a,b)
  return inner
```

```python
@decor1
def div(a,b):
  return a/b
```

```python
div(7,2)
```

## 2. Decorators

A **decorator** is a design pattern in Python that allows you to add new functionality to an existing object without modifying its structure. Decorators are essentially functions that take another function as an argument, extend its behavior, and return a new function. They are often used for logging, timing, authentication, caching, and more.

```python
# Example of a Decorator

def my_decorator(func):
  def wrapper(*args, **kwargs):
    print("Something is happening before the function is called.")
    result = func(*args, **kwargs) # Call the original function
    print("Something is happening after the function is called.")
    return result
  return wrapper

@my_decorator
def say_hello(name):
  print(f"Hello, {name}!")
  return f"Hello, {name} from say_hello"

# Call the decorated function
output = say_hello("Alice")
print(f"Function output: {output}")

print("\n--- Another decorated function ---")

@my_decorator
def calculate_sum(a, b):
  print(f"Calculating sum of {a} and {b}")
  return a + b

sum_result = calculate_sum(10, 20)
print(f"Sum result: {sum_result}")
```
