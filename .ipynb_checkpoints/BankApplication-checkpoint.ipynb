```python
%%writefile password_validate.py

def password_validation(password):
    if len(password) >= 8:
        lower = False
        upper = False
        digit = False
        special = False
        for i in password:
            if i.isupper():
                upper = True
            elif i.islower():
                lower = True
            elif i.isdigit():
                digit = True
            elif not i.isalnum():
                special = True
        if all([lower, upper, digit, special]):
            return True
        else:
            return False
    else:
        return False
```

```python
from password_validate import password_validation

data = dict()

def signup():
    user_name = input('Enter user name: ')
    password = input('Enter password: ')
    if password_validation(password):
        print("Initial amount of 2000 is required to open an account...")
        choice = input('Enter your choice(y/n): ').lower()
        if choice == 'y':
            balance = 2000
            if data:
                acc_no = list(data.keys())[-1] + 1
                data[acc_no] = [user_name, password, balance]
                print(f"Your account no. is {acc_no}")
                
            else:
                data[1001] = [user_name, password, balance]
                print(f"Your account is created successfully...")
                print(f"Your account no. is 1001")
        else:
            print("Initial amount is required to open the account...\nThank you for your time...")
    else:
        print("Please check your password...")


def login():
    acc_no = int(input('Enter acc number: '))
    password = input('Enter your password: ')
    
    if acc_no in data and data[acc_no][1] == password:
        print("1.Credit\n2.Debit\n3.Change Password\n4.Check balance\n5.Logout")
        while True:
            query = int(input('Enter query no: '))
        
            if query == 1:
                credit = int(input('Enter your credit amount: '))
                data[acc_no][2] += credit
                print(f"Amount credited successfully...")
            
            elif query == 2:
                debit = int(input('Enter your debit amount: '))
                if data[acc_no][2] - debit >= 2000:
                    data[acc_no][2] -= debit
                    print(f"Amount debited successfully...")
                    print(f"Remining balance: {data[acc_no][2]}")
                else:
                    print("Not enough balance...")
            elif query == 3:
                old_password = input('Enter old password: ')
                if old_password == data[acc_no][1]:
                    new_password = input('Enter new password: ')
                    if password_validation(new_password):
                        data[acc_no][1] = new_password
                    else:
                        print(f"Invalid password...")
                else:
                    print(f"Wrong old password...")
            elif query == 4:
                print(f"Current balance: {data[acc_no][2]}")
            elif query == 5:
                break
    else:
        print("Invalid account no or password...")

def forget_password():
    pass


    
```

```python
while True:
    print("1.Login\n2.Signup\n3.Forget Password\n4.Exit")
    query = int(input('Enter your query: '))
    if query == 1:
        login()
    elif query == 2:
        signup()
    elif query == 3:
        forget_password()
    elif query == 4:
        break
```
