# Recursion: Function Calling itself

```
def func():
    func()
```

### Factorial using loop

````python
fact = 1
num = 5
for i in range(1, num + 1):
    fact = fact * i
print(fact)
````

## Factorial using Recursion

````python
def fact(n):
    if n == 1:
        return 1
    n = n * fact(n-1)
    return n
````

````python
fact(5)
````
