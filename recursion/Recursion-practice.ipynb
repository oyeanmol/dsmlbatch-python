```python
def sum_number(n):
    if n == 0:
        return 0
    return n + sum_number(n - 1)
```

# Factorial

!5 = 5 x 4 x 3 x 2 x 1 = 120

```python
n = int(input("Enter your number: "))

fact = 1

# using loop

for i in range(1, n + 1):
    fact = fact * i

#                1  *    2
# 2

#                 2 * 3
# 6

#                 6 * 4
# 24
#                 24 * 5
# 120

print(f"Factorial of {n}: {fact}")
```

# Using Recursion

```python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)
```

```python
print(factorial(5))
```

```python
def greet(n):
    if n == 0:
        return
    print("Anmol")
    greet(n - 1)
```

```python
greet(5)
```
