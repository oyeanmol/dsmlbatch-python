# JSON - Universal Serializer

            
          Serializers --->  json.dump()
                            json.dumps()

          Deserializers --> json.load()
                            json.loads()

                            

```python
import json
```

```python
ages = [20, 21 ,25 ,27, 26, 30]
```

```python
print(type(ages))
```

```python
f = open("newdata.json", 'w')
```

```python
json.dump(ages, f)
```

```python
f.close()
```

```python
f = open("newdata.json", 'r')
data2 = json.load(f)
```

```python
print(data2)
```

```python
print(type(data2))
```

```python
with open("jsonfile.json", 'w') as file:
    json.dump(ages, file)
```

```python
f = open("jsonfile.json", 'r')
```

```python
age = json.load(f)
```

```python
print(age)
```
