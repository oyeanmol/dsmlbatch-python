# Dunder Methods in Python OOPs

## Complete Jupyter Notebook

**Dunder = Double Underscore**

Dunder methods are special methods in Python whose names start and end with double underscores (`__`).

Examples:

```python
__init__
__str__
__len__
__add__
__eq__
__lt__
__call__
```

They are also called **special methods** or **magic methods**.

> The most important idea: you usually do not call dunder methods directly. Python calls them automatically when you perform certain operations on an object.


## 1. Why Do We Need Dunder Methods?

Suppose we create a `Student` class:

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
```

When we print an object, Python does not automatically know the user-friendly representation we want.

Dunder methods allow us to define how our custom objects behave with Python's built-in syntax.


````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

s1 = Student("Anmol", 85)
s1
````

The output is typically similar to:

```text
<__main__.Student object at 0x...>
```

We can customize this behavior using `__str__()`.


````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __str__(self):
        return f"{self.name} - {self.marks}"

s1 = Student("Anmol", 85)

print(s1)
````

Conceptually:

```python
print(s1)
```

uses:

```python
s1.__str__()
```

This is the basic idea behind dunder methods.

# 3. `__init__()` — Constructor

`__init__()` is used to initialize an object after the instance has been created.

When we write:

```python
s1 = Student("Anmol", 85)
```

Python invokes the initialization process for that object.

Example:


````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

s1 = Student("Anmol", 85)

print(s1.name)
print(s1.marks)
````

Flow:

```text
Student(...)
     ↓
__init__()
     ↓
Object attributes initialized
```

> More precisely, `__new__()` creates/returns the instance, while `__init__()` initializes it. We will see this later.

# 5. `__repr__()` — Developer-Friendly Representation

`__repr__()` is generally intended to provide an unambiguous, developer-oriented representation of an object.

Example:


````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __repr__(self):
        return f"Student('{self.name}', {self.marks})"

s1 = Student("Anmol", 90)

print(repr(s1))
````

## `__str__()` vs `__repr__()`

| Method | Purpose |
|---|---|
| `__str__()` | User-friendly representation |
| `__repr__()` | Developer/debugging representation |

Example:


````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __str__(self):
        return f"{self.name} scored {self.marks}"

    def __repr__(self):
        return f"Student('{self.name}', {self.marks})"

s1 = Student("Anmol", 90)

print(s1)
print(repr(s1))
````

# 6. `__len__()` — `len(object)`

`__len__()` defines what should happen when `len()` is used on our object.


````python
class Course:
    def __init__(self, students):
        self.students = students

    def __len__(self):
        return len(self.students)

course = Course(["A", "B", "C", "D"])

print(len(course))
````

Conceptually:

```python
len(course)
```

uses:

```python
course.__len__()
```


# 7. `__add__()` — `+` Operator

`__add__()` lets us define what `+` should mean for our custom objects.

This is an example of **operator overloading**.


````python
class Student:
    def __init__(self, marks):
        self.marks = marks

    def __add__(self, other):
        return self.marks + other.marks

s1 = Student(80)
s2 = Student(90)

print(s1 + s2)
````

Conceptually:

```python
s1 + s2
```

uses:

```python
s1.__add__(s2)
```


# 8. `__sub__()` — `-` Operator


````python
class Number:
    def __init__(self, value):
        self.value = value

    def __sub__(self, other):
        return self.value - other.value

n1 = Number(50)
n2 = Number(20)

print(n1 - n2)
````

# 9. `__mul__()` — `*` Operator


````python
class Number:
    def __init__(self, value):
        self.value = value

    def __mul__(self, other):
        return self.value * other.value

n1 = Number(10)
n2 = Number(5)

print(n1 * n2)
````

# 10. `__truediv__()` — `/` Operator


````python
class Number:
    def __init__(self, value):
        self.value = value

    def __truediv__(self, other):
        return self.value / other.value

n1 = Number(20)
n2 = Number(5)

print(n1 / n2)
````

# 11. `__eq__()` — `==`

`__eq__()` defines equality comparison between objects.

By default, two separate objects are not generally equal merely because their attributes contain the same values.

We can define our own equality logic:


````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __eq__(self, other):
        return self.name == other.name and self.marks == other.marks

s1 = Student("Anmol", 90)
s2 = Student("Anmol", 90)

print(s1 == s2)
````

Conceptually:

```python
s1 == s2
```

uses:

```python
s1.__eq__(s2)
```

# 13. Example of `__lt__()` and `__gt__()`

Let's compare students based on marks.


````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __lt__(self, other):
        return self.marks < other.marks

    def __gt__(self, other):
        return self.marks > other.marks

s1 = Student("Anmol", 80)
s2 = Student("Rahul", 90)

print(s1 < s2)
print(s1 > s2)
````

# 14. `__contains__()` — `in`

`__contains__()` lets us define how membership testing works with our object.


````python
class Course:
    def __init__(self, subjects):
        self.subjects = subjects

    def __contains__(self, item):
        return item in self.subjects

course = Course(["Python", "Django", "SQL"])

print("Python" in course)
print("Java" in course)
````

Conceptually:

```python
"Python" in course
```

uses:

```python
course.__contains__("Python")
```


# 15. `__getitem__()` — `object[index]`

`__getitem__()` allows an object to support indexing or key-based access.


````python
class StudentList:
    def __init__(self, students):
        self.students = students

    def __getitem__(self, index):
        return self.students[index]

students = StudentList(["Anmol", "Rahul", "Aman"])

print(students[0])
print(students[1])
````

Conceptually:

```python
students[0]
```

uses:

```python
students.__getitem__(0)
```


# 16. `__setitem__()` — `object[index] = value`

`__setitem__()` controls assignment using indexing.


````python
class StudentList:
    def __init__(self, students):
        self.students = students

    def __getitem__(self, index):
        return self.students[index]

    def __setitem__(self, index, value):
        self.students[index] = value

students = StudentList(["Anmol", "Rahul", "Aman"])

students[1] = "Rohit"

print(students[1])
print(students.students)
````

# 17. `__delitem__()` — `del object[index]`

`__delitem__()` controls deletion using indexing.


````python
class StudentList:
    def __init__(self, students):
        self.students = students

    def __delitem__(self, index):
        del self.students[index]

students = StudentList(["Anmol", "Rahul", "Aman"])

del students[1]

print(students.students)
````

# 18. `__call__()` — Make an Object Callable

Normally, functions can be called:

```python
greet()
```

But we can make an object itself callable using `__call__()`.


````python
class Calculator:
    def __call__(self, a, b):
        return a + b

calc = Calculator()

print(calc(10, 20))
````

Conceptually:

```python
calc(10, 20)
```

uses:

```python
calc.__call__(10, 20)
```


## Practical `__call__()` Example

A class can store configuration and then be used like a function.


````python
class Multiplier:
    def __init__(self, factor):
        self.factor = factor

    def __call__(self, number):
        return number * self.factor

double = Multiplier(2)
triple = Multiplier(3)

print(double(10))
print(triple(10))
````

# 19. `__bool__()` — `bool(object)`

`__bool__()` defines whether an object should evaluate to `True` or `False`.


````python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def __bool__(self):
        return self.balance > 0

account = BankAccount(5000)

if account:
    print("Account has balance")
else:
    print("Account is empty")
````

Conceptually:

```python
bool(account)
```

uses:

```python
account.__bool__()
```


# 20. `__iter__()` — Make an Object Iterable

`__iter__()` allows an object to provide an iterator for operations such as `for` loops.


````python
class Course:
    def __init__(self, subjects):
        self.subjects = subjects

    def __iter__(self):
        return iter(self.subjects)

course = Course(["Python", "Django", "SQL"])

for subject in course:
    print(subject)
````

# 21. `__next__()` — Iterator's Next Item

An iterator commonly implements:

```python
__iter__()
__next__()
```

Example:


````python
class Counter:
    def __init__(self, max_value):
        self.max_value = max_value
        self.current = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= self.max_value:
            value = self.current
            self.current += 1
            return value

        raise StopIteration

counter = Counter(3)

for number in counter:
    print(number)
````

Iterator flow:

```text
for loop
   ↓
__iter__()
   ↓
__next__()
   ↓
1
   ↓
__next__()
   ↓
2
   ↓
__next__()
   ↓
3
   ↓
__next__()
   ↓
StopIteration
```


# 22. `__enter__()` and `__exit__()` — `with`

These methods implement the context manager protocol.

They are useful for resources such as files, database connections, locks, and other setup/cleanup operations.


````python
class Demo:
    def __enter__(self):
        print("Entering")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print("Exiting")

with Demo() as obj:
    print("Inside block")
````

Flow:

```text
with Demo()
     ↓
__enter__()
     ↓
Inside block
     ↓
__exit__()
```

A major real-world example is:

```python
with open("data.txt", "r") as file:
    data = file.read()
```


# 23. `__new__()` vs `__init__()`

This is an important advanced concept.

- `__new__()` is responsible for creating/returning the instance.
- `__init__()` initializes an already-created instance.

Example:


````python
class Student:
    def __new__(cls):
        print("__new__ called")
        return super().__new__(cls)

    def __init__(self):
        print("__init__ called")

s1 = Student()
````

Flow:

```text
Student()
   ↓
__new__()
   ↓
Object created/returned
   ↓
__init__()
   ↓
Object initialized
```

So remember:

```text
__new__()  → creates/returns instance
__init__() → initializes instance
```


# 24. `__del__()` — Object Finalization

Python provides `__del__()` as a finalizer hook that may run when an object is being finalized.

Example:


````python
class Demo:
    def __init__(self):
        print("Object created")

    def __del__(self):
        print("Object finalized")

obj = Demo()

del obj
````

### Important note

Do not use `__del__()` for important resource cleanup because its execution timing is not something you should rely on for deterministic cleanup.

Prefer:

- context managers (`with`)
- explicit cleanup methods

for resources that require reliable cleanup.


# 25. `__dict__` — Important Distinction

`__dict__` is **not a dunder method**. It is generally an attribute containing an object's namespace.

Example:


````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

s1 = Student("Anmol", 90)

print(s1.__dict__)
````

You get something similar to:

```python
{'name': 'Anmol', 'marks': 90}
```

Notice the difference:

```python
__dict__     # attribute
__str__()    # method
```


# 26. Operator Overloading

Operator overloading means giving operators such as:

```text
+  -  *  /  ==  <  >
```

a customized meaning for our own objects.

For example:

```python
10 + 20
```

means numeric addition.

But:

```python
student1 + student2
```

can be defined to mean something else, such as combining marks.

Similarly:

```python
student1 > student2
```

can compare their marks.

Dunder methods make this possible.


# 27. Real-World Example — Product

Let's create a `Product` class where:

- `print(product)` displays the product
- `product1 + product2` gives combined price
- `product1 < product2` compares prices


````python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        return f"{self.name} - ₹{self.price}"

    def __add__(self, other):
        return self.price + other.price

    def __lt__(self, other):
        return self.price < other.price

p1 = Product("Keyboard", 1500)
p2 = Product("Mouse", 800)

print(p1)
print(p2)

print("Combined price:", p1 + p2)
print("Is p1 cheaper than p2?", p1 < p2)
````

# 28. Complete Example — Student Class

This example combines several important dunder methods.


````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    # print(object)
    def __str__(self):
        return f"{self.name}: {self.marks}"

    # repr(object)
    def __repr__(self):
        return f"Student('{self.name}', {self.marks})"

    # object1 + object2
    def __add__(self, other):
        return self.marks + other.marks

    # object1 == object2
    def __eq__(self, other):
        return self.marks == other.marks

    # object1 > object2
    def __gt__(self, other):
        return self.marks > other.marks

    # len(object)
    def __len__(self):
        return len(self.name)

    # object()
    def __call__(self):
        print(f"{self.name} is called")


s1 = Student("Anmol", 90)
s2 = Student("Rahul", 80)

print(s1)
print(repr(s1))

print(s1 + s2)

print(s1 == s2)

print(s1 > s2)

print(len(s1))

s1()
````

# 32. Final Takeaway

Dunder methods are Python's mechanism for defining **how custom objects interact with Python's built-in syntax and operations**.

Instead of writing special functions outside the class, we define the behavior inside the class:

```python
class MyClass:

    def __str__(self):
        ...

    def __add__(self, other):
        ...

    def __eq__(self, other):
        ...

    def __len__(self):
        ...
```

Then Python automatically uses these methods when we write familiar operations:

```python
print(obj)
obj1 + obj2
obj1 == obj2
len(obj)
obj[0]
item in obj
obj()
```

### Core idea

```text
Python syntax
     ↓
Special/Dunder Method
     ↓
Your custom object behavior
```

Once you understand this connection, operator overloading, iterable objects, callable objects, context managers, and many Python frameworks become much easier to understand.
