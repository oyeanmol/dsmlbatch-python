# Inheritance in Python OOPs

Inheritance is an Object-Oriented Programming concept in which a **child class** acquires the attributes and methods of a **parent class**.

### Why use inheritance?
- Code reusability
- Reduces duplicate code
- Makes programs easier to maintain
- Helps create relationships between classes

### Basic Syntax

```python
class Parent:
    # parent class members
    pass

class Child(Parent):
    # child class members
    pass
```

**Terminology:**
- Parent class → Base class / Superclass
- Child class → Derived class / Subclass


## 1. Basic Inheritance Example

A child class can directly use methods defined in its parent class.


````python
class Animal:
    def eat(self):
        print("Animal is eating")


class Dog(Animal):
    def bark(self):
        print("Dog is barking")


dog = Dog()

dog.eat()   # inherited method
dog.bark()  # child class method
````

## 2. Single Inheritance

**Single inheritance** occurs when one child class inherits from one parent class.

```text
Parent
   ↓
Child
```


````python
class Vehicle:
    def start(self):
        print("Vehicle started")


class Car(Vehicle):
    def drive(self):
        print("Car is driving")


car = Car()

car.start()
car.drive()
````

### Single Inheritance with `__init__`

A child class can have its own constructor while also using functionality from the parent class.


````python
class Person:
    def __init__(self, name):
        self.name = name

    def show_name(self):
        print("Name:", self.name)


class Student(Person):
    def study(self):
        print(self.name, "is studying")


student = Student("Anmol")

student.show_name()
student.study()
````

## 3. Multiple Inheritance

**Multiple inheritance** occurs when one child class inherits from more than one parent class.

```text
Parent1     Parent2
    \         /
       Child
```

Python allows a class to inherit from multiple classes.


````python
class Father:
    def driving(self):
        print("Driving skill")


class Mother:
    def cooking(self):
        print("Cooking skill")


class Child(Father, Mother):
    def playing(self):
        print("Playing")


child = Child()

child.driving()
child.cooking()
child.playing()
````

### Multiple Inheritance with Constructors

When multiple parent classes have constructors, the order in which they are called matters. `super()` follows Python's **Method Resolution Order (MRO)**.


````python
class A:
    def __init__(self):
        print("A constructor")


class B:
    def __init__(self):
        print("B constructor")


class C(A, B):
    def __init__(self):
        A.__init__(self)
        B.__init__(self)
        print("C constructor")


obj = C()
````

## 4. Multilevel Inheritance

**Multilevel inheritance** creates a chain of inheritance.

```text
Grandparent
     ↓
   Parent
     ↓
   Child
```

The last child can access inherited members from all classes in the chain.


````python
class Grandfather:
    def property(self):
        print("Grandfather's property")


class Father(Grandfather):
    def car(self):
        print("Father's car")


class Son(Father):
    def bike(self):
        print("Son's bike")


son = Son()

son.property()
son.car()
son.bike()
````

## 5. Method Resolution Order (MRO)

**MRO** determines the order in which Python searches for a method or attribute in an inheritance hierarchy.

You can see the MRO using:

```python
ClassName.mro()
```

or:

```python
ClassName.__mro__
```


````python
class A:
    def show(self):
        print("A")


class B(A):
    def show(self):
        print("B")


class C(A):
    def show(self):
        print("C")


class D(B, C):
    pass


print(D.mro())
````

### Understanding the MRO

For the example above, Python follows an order similar to:

```text
D → B → C → A → object
```

So if `D` does not contain `show()`, Python checks `B`, then `C`, then `A`.

This is especially important in **multiple and hybrid inheritance**.


````python
class A:
    def show(self):
        print("A")


class B(A):
    def show(self):
        print("B")


class C(A):
    def show(self):
        print("C")


class D(B, C):
    pass


obj = D()

obj.show()

print("MRO:")
for cls in D.mro():
    print(cls.__name__)
````

## 6. `super()` in Python

`super()` is used to access a method or constructor of the next class in the MRO.

It is commonly used to:
- Call a parent class constructor
- Reuse parent class methods
- Work correctly with multiple inheritance


````python
class Person:
    def __init__(self, name):
        self.name = name


class Student(Person):
    def __init__(self, name, course):
        super().__init__(name)
        self.course = course

    def show(self):
        print("Name:", self.name)
        print("Course:", self.course)


student = Student("Anmol", "Python")

student.show()
````

### `super()` with a Parent Method


````python
class Animal:
    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):
    def sound(self):
        super().sound()
        print("Dog barks")


dog = Dog()
dog.sound()
````

## 7. `super()` and Multiple Inheritance

`super()` does **not simply mean "call my direct parent"**.

It follows the class's **MRO** and calls the next class in that order.

This makes `super()` useful for cooperative multiple inheritance.


````python
class A:
    def show(self):
        print("A")
        super().show()


class B(A):
    def show(self):
        print("B")
        super().show()


class C(A):
    def show(self):
        print("C")
        super().show()


class D(B, C):
    def show(self):
        print("D")
        super().show()


class End:
    def show(self):
        print("End")


# Connect the chain by using End as the final class
# A small practical version is shown below.
````

### A Complete Cooperative Multiple-Inheritance Example

Here, every class uses `super()`, so Python follows the MRO instead of manually calling a specific parent.


````python
class A:
    def show(self):
        print("A")
        super().show()


class B(A):
    def show(self):
        print("B")
        super().show()


class C(A):
    def show(self):
        print("C")
        super().show()


class D(B, C):
    def show(self):
        print("D")
        super().show()


class End:
    def show(self):
        print("End")


# Recreate the hierarchy with End at the end of the MRO.
class A2(End):
    def show(self):
        print("A")
        super().show()


class B2(A2):
    def show(self):
        print("B")
        super().show()


class C2(A2):
    def show(self):
        print("C")
        super().show()


class D2(B2, C2):
    def show(self):
        print("D")
        super().show()


obj = D2()

print("MRO:", [cls.__name__ for cls in D2.mro()])
print("\nMethod calls:")
obj.show()
````

## 8. Inheritance Types at a Glance

| Type | Description | Example |
|---|---|---|
| **Single** | One parent → one child | `Child(Parent)` |
| **Multiple** | Multiple parents → one child | `Child(A, B)` |
| **Multilevel** | Parent → child → grandchild | `C(B)`, `B(A)` |
| **Hierarchical** | One parent → multiple children | `B(A)`, `C(A)` |
| **Hybrid** | Combination of inheritance types | Multiple + hierarchical, etc. |

> This notebook focuses mainly on **Single, Multiple, Multilevel, MRO, and `super()`**.


## 9. Important Points to Remember

1. A child class can access inherited methods and attributes from its parent.
2. Python supports multiple inheritance.
3. **MRO** decides the order in which Python searches classes.
4. Use `ClassName.mro()` to inspect the MRO.
5. `super()` follows the MRO.
6. `super()` is preferred over hard-coding parent class names when designing cooperative inheritance.
7. `object` is the ultimate base class for normal Python classes.
