# ***`OOPS`***
    - Object oriented programming system / structure

## `Two type of programming`
    - Procedural programming
        - using function
    - Object Oriented Programming
        - using oops

## `Class`
    - class is a blueprint / template / structure
    - Example
        - Human is a class
        - Universe is a class

## `Class consist of majorly two things`
    - Methods / Behaviour 
        - Functions that are defined under a class
        - Example
            - List is a class
            - Append , Extend is method of list class
            - Human is a class
            - Sing() - method
            - EAT() - method
            - Study() - method
            - Listen() - Method
    - Attribute
        - Variables that are declared inside a class
        - Human >  class
        - Name > Attr
        - Age > Attr


## `Two type of class`
    - Built-In-Class
        - Class that are already predefined by python
        - Example
            - list
            - int
            - str
            - dict
            - set

    - User-Defined-Class
        - Class which are defined by user


# `Object`
    - Object can be anything
    - object is also called member / Instance
    - A object can use properties of a class from which that he belongs

## `Example`
    - Class - Human
    - Attribute - Age , Address
    - Methods - sing() , eat() , listen() , play()

    - Simran - object - class Human
    - Now simran can use any methods of class Human
    - Becoz she is now became member of class Human


## `Python Example`
    - class - List
    - Method - Append() , count() , extend() , insert()
    - ls became member of list class
    - Now ls can access any method of class list

## `Advantage of oops`
    - Modularity
    - Flexible
    - Data security
    - Data Hide

## `Four Pillars of oops`
    - Abstraction
        - Hiding the data 
    - Inheritance
        - Inherit/Derive property from any other class
    - Polymorphism
        - poly > Many
        - morphism > forms
        - One thing which have many forms
    - Encapsulation
        - Binding all the data with the security into a class

# Constructor in Python

Constructor ka main purpose hota hai object ko initialize karna — yani object create hote hi uske initial data ko set karna.

Python mein commonly `__init__()` ko constructor kaha jata hai.

```python
class Human:
    def __init__(self):
        pass
```

```python
class Student:
    lab = "Lab 3"
    # def __init__(self):
    #     print("Student object created")

    def __init__(self, name):
        self.name = name
        print("Name: ", self.name)
    def show(self):
        print(self)

# krish = Student()

# krish.lab
```

```python
s1 = Student("Krish")
print(s1)
s1.show()
```

# Constructor with parameter:

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show(self):
        print(self.name)
        print(self.age)


s1 = Student("Anmol", 22)

s1.show()
```

 - def __init__(self, name, age):
 - self → current object ka reference
 - name, age → parameters
 - "Anmol", 22 → arguments
 - self.name, self.age → instance variables

# Self

Reference of the Object itself

```python
class New:

    def show(self):
        print("self = ", self)
```

```python
s1 = New()  # creating an Object of Class New
print("s1 = ", s1) # Printing Object itself
s1.show() # Accessing Show() Method using object
```

Types of Variables:

1. Instance Variable
2. Class Variable

```python
class Factory:
    # Class Varible
    company = "BMW"
    color = "Black"
    def __init__(self, model, engine):
        # Instance Variables
        self.CarKaModel = model
        self.CarKaEngine = engine
        color = "Black"
```

```python
car1 = Factory("M5", "V8")
```

```python
car1.__dict__
```

```python
car2 = Factory("M3", "V12")
```

```python
car1.company
```

```python
car2.company
```

```python
car1.CarKaEngine
```

```python
car2.CarKaEngine
```

```python
print(car1.color)
```

```python
car1.CarKaModel
```

```python
car3 = Factory("M8", "V6")
```

```python
car3.CarKaModel
```
