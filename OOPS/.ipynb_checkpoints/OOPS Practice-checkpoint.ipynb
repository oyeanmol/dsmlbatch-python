```python
# Parent/base
class Parent:
    def method1(self):
        print("This is method-1")

# Child Class / Derived Class
class Child(Parent):
    def method2(self):
        print("This is method-2")
```

```python
a = Child()
```

```python
a.method2()
```

```python
a.method2()
```

```python
a.method1()
```

```python
class Human:
    def __init__(self, name):
        self.name = name
        print(f"name: {self.name}")

    def show_name(self):
        print("Name:", self.name)


class Student(Human):
    def __init__(self, other):
        self.other = other
        print(f"other: {self.other}")
    def study(self):
        print(self.name, "is studying")
```

```python
pradeep = Student("Pradeep")
```

```python
pradeep.show_name()
```

```python
pradeep.study()
```

```python
class RamSir:
    def aptitude(self):
        print("Aptitude Teacher: Ram Sir")

class ShyamSir:
    def devops(self):
        print("Dev Ops Teacher")
        
class AnmolSir:
    def python(self):
        print("Anmol Sir")

class Student(RamSir, ShyamSir, AnmolSir):
    def __init__(self, name):
        self.name = name
        print(f"{self.name}")
    def study(self):
        print("Who taught you?")
```

```python
tasleem = Student()
```

```python
pradeep.python()
```

```python
tasleem.aptitude()
```

```python
# Multiple Inheritance with constuctors
class A:
    def __init__(self, name):
        print("A constructor")


class B:
    def __init__(self, age):
        print("B constructor")


class C(A, B):
    # MRO
    def __init__(self):
        print("C constructor")
        A.__init__(self)
        B.__init__(self)
```

```python
obj1 = C()
```

```python
class A:
    def __init__(self, name):
        self.name = name
        print("A constructor")
    def method1(self):
        print("Method 1")


class B:
    def __init__(self, age):
        self.age = age
        print("B constructor")


class C(A, B):
    def __init__(self, salary, name, age):
        self.salary = salary
        # Explicit Calling the constructors
        A.__init__(self, name)
        B.__init__(self,age)
        print("C Constructor")

```

```python
ob1 = C(50000, "Anmol", 26)
```

```python
print(obj1.salary)
```

```python
class Time:

    def __init__(self, hours, minutes):
        self.hours = hours
        self.minutes = minutes

    def addTime(self, t1, t2):
        total_minutes = (t1.hours * 60 + t1.minutes) + (t2.hours * 60 + t2.minutes)
        self.hours = total_minutes // 60
        self.minutes = total_minutes % 60

    def displayTime(self):
        print(f"{self.hours} hr {self.minutes} min")

    def displayMinutes(self):
        total_minutes = self.hours * 60 + self.minutes
        print(total_minutes, " minutes")
```

```python
t1 = Time(2, 30)
t2 = Time(1, 40)

result = Time(0, 0)
result.addTime(t1, t2)
result.displayTime()
result.displayMinutes()
```

```python
class Person:
    def __init__(self, f_name, l_name, phone, email):
        self.name = f_name
        self.lname = l_name
        self.phone = phone
        self.email = email
    def __str__(self):
        return f"Name: {self.name} {self.lname}\nPhone: {self.phone}\nEmail: {self.email}"
```

```python
a = Person("Anmol", "Shah", 8000348124, ["anmol123@gmail.com", "anmolxyz@gmail.com"])
```

```python
print(a)
```

```python
class A:
    def show(self):
        print("A")


class B(A):
    def show(self):
        print("B")
        A.show(self)


class C(A):
    def show(self):
        print("C")
        super().show()


class D(B, C):
    def show(self):
        print("D")
        super().show()
```

```python
a = D()
```

```python
a.show()
```
