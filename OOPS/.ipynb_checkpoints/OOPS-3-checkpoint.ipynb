# Constructor in Python

Constructor ka main purpose hota hai object ko initialize karna — yani object create hote hi uske initial data ko set karna.

Python mein commonly `__init__()` ko constructor kaha jata hai.

````python
class Student:
    lab = "Lab 3"
    # def __init__(self):
    #     print("Student object created")

    def __init__(self, name):
        self.name = name
        print("Name: ", self.name)
    def show(self):
        print(self)

# krish = Student()

# krish.lab
````

````python
s1 = Student("Krish")
print(s1)
s1.show()
````

# Constructor with Parameters

````python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show(self):
        print(self.name)
        print(self.age)


s1 = Student("Anmol", 22)

s1.show()
````

````python
# def __init__(self, name, age):
# self → current object ka reference
# name, age → parameters
# "Anmol", 22 → arguments
# self.name, self.age → instance variables
````

# Destructor in Python

Destructor ka purpose object ke cleanup/finalization se related hota hai

using: `__del__()`

````python
class Student:

    def __init__(self):
        print("Object created")

    def __del__(self):
        print("Object destroyed")


s1 = Student()

del s1
````

## del kya karta hai?
`
del s1
`
del object ko directly "destroy" karne ka guaranteed command nahi hai.

It deletes the reference s1.

If that was the last reference to the object, the object can become unreachable and Python may then finalize it, potentially calling __del__().

````python
class Student:

    def __del__(self):
        print("Destructor called")


s1 = Student()

del s1
````

Python mein __del__() ka exact execution time guaranteed nahi hota.

Python ka garbage collector / reference counting implementation object ko kab clean up karega, ye situation aur Python implementation par depend kar sakta hai.

# Method
A method is a function defined inside a class.

````python
class Student:

    def study(self):
        print("Student is studying")
````

````python
# study()

s1 = Student()

s1.study()
````

# Function vs Method
## Function
```python
def add(a, b):
    return a + b
```
##  Method
```python
class Calculator:

    def add(self, a, b):
        return a + b

```
The main difference is where it is defined:

```
Function → defined independently

Method → defined inside a class
```

# Types of Methods in Python

For Python OOP, the three main types are:
```
Methods
│
├── Instance Method
├── Class Method
└── Static Method
```
There are also special/dunder methods.

````python
# Instance Method

#An instance method works with a particular object/instance.

#It normally takes self as its first parameter.

#Example
class Student:

    def show(self):
        print("Student details")


s1 = Student()
s1.show()
````

Here:
```
def show(self):
```
is an instance method.

# Question

Make a class "Person" asking for name and age and then create 2 objects and then compare both the objects using age of both objects. for that create a method called compare that compares 2 objects by their age.

````python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show(self):
        print("Name:", self.name)
        print("Age:", self.age)


s1 = Student("Anmol", 22)

s1.show()
````

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        
    def compare(self, other):
        if self.age > other.age:
            return True
        else:
            return False
````

````python
p1 = Person("Anmol", 24)
````

````python
p2 = Person("Ashish", 25)
````

````python
if p1.compare(p2):
    print(f"{p1.name} is younger than {p2.name}")
else:
    print(f"{p2.name} is younget than {p1.name}")
````

# Instance Method with Instance Variables

This is the most important use.

````python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show(self):
        print("Name:", self.name)
        print("Age:", self.age)


s1 = Student("Anmol", 22)

s1.show()
````

So:

Instance method → works with instance/object data → uses self.

# Class Method

A class method works with the class rather than a particular object.

It is created using:

````python
@classmethod
````

````python
class Student:

    college = "GRRAS"

    @classmethod
    def show_college(cls):
        print(cls.college)


Student.show_college()
````

# Example: Student Class

````python
class Student:

    school = "ABC School"   # Class Variable

    def __init__(self, name, age):
        self.name = name    # Object Variable
        self.age = age

    # Instance Method
    def show_student(self):
        print("Name:", self.name)
        print("Age:", self.age)

    # Class Method
    @classmethod
    def change_school(cls, new_school):
        cls.school = new_school


# Creating Objects
student1 = Student("Rahul", 20)
student2 = Student("Aman", 21)

# Object method
student1.show_student()
student2.show_student()

# Class Method
print("Old School:", Student.school)

Student.change_school("XYZ School")

print("New School:", Student.school)
````
