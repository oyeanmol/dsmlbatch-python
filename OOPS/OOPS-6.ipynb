## 1. What is Polymorphism?

**Polymorphism** means **"many forms."**

In OOP, polymorphism allows the **same interface, method, or operation** to behave differently depending on the object involved.

### Simple idea

```text
Same method / operation
        ↓
Different objects
        ↓
Different behavior
```

For example, different animals can have the same `sound()` method, but each animal produces a different sound.

```text
Dog → Bark
Cat → Meow
Cow → Moo
```

The method name is the same, but the behavior is different.

### Main forms of polymorphism commonly discussed in Python

1. **Method Overriding**
2. **Duck Typing**
3. **Operator Overloading**
4. **Method Overloading using default/variable arguments**


## 2. Method Overriding

Method overriding occurs when a **child class provides its own implementation of a method that already exists in the parent class**.

This is one of the most important examples of runtime polymorphism in Python.

### Syntax

```python
class Parent:
    def method(self):
        # parent implementation
        pass

class Child(Parent):
    def method(self):
        # child implementation
        pass
```


````python
class Animal:
    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):
    def sound(self, name):
        print("Dog barks")


class Cat(Animal):
    def sound(self):
        print("Cat meows")


dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
````

### Explanation

Here, `Dog` and `Cat` both inherit from `Animal`.

Both classes override the `sound()` method:

- `Dog.sound()` → `Dog barks`
- `Cat.sound()` → `Cat meows`

The same method name `sound()` produces different behavior depending on the object.

## 3. Polymorphism with a Common Interface

A major advantage of polymorphism is that we can write a function that works with **different types of objects**.

We do not need a separate function for every class.


````python
class Dog:
    def sound(self):
        print("Bark")


class Cat:
    def sound(self):
        print("Meow")


def make_sound(animal):
    animal.sound()


dog = Dog()
cat = Cat()

make_sound(dog)
make_sound(cat)
````

### Why is this polymorphism?

The function `make_sound()` does not need to know whether it received a `Dog` or a `Cat`.

It simply expects the object to provide a `sound()` method.

```text
make_sound(Dog) → Bark
make_sound(Cat) → Meow
```

This makes the code flexible and reusable.

## 4. Duck Typing

Python uses a concept called **Duck Typing**.

The idea is commonly summarized as:

> "If it behaves like a duck, treat it like a duck."

Python generally focuses on **what an object can do**, rather than strictly checking its class.

For example, if an object has a `sound()` method, our function can call it regardless of the object's class.


````python
class Dog:
    def sound(self):
        print("Bark")


class Cat:
    def sound(self):
        print("Meow")


class Cow:
    def sound(self):
        print("Moo")


def make_sound(animal):
    animal.sound()


make_sound(Dog())
make_sound(Cat())
make_sound(Cow())
````

### Important Point

Notice that `Dog`, `Cat`, and `Cow` do not need to inherit from a common parent class.

As long as the object provides the required `sound()` method, `make_sound()` can use it.

This is **duck typing**.

## 5. Polymorphism with Built-in Functions

Python's built-in functions also demonstrate polymorphism.

For example, `len()` works with many different objects:

- String
- List
- Tuple
- Dictionary
- Set
- Other objects that implement the required protocol


````python
text = "Python"
numbers = [10, 20, 30, 40]
data = (1, 2, 3)

print(len(text))
print(len(numbers))
print(len(data))
````

The same function `len()` behaves differently depending on the object passed to it.

```text
len("Python")      → 6
len([10, 20, 30])  → 3
len((1, 2, 3))     → 3
```

This is another practical example of polymorphic behavior in Python.

## 6. Operator Overloading

Python allows operators such as:

```text
+   -   *   /   ==   <   >
```

to behave differently for different data types.

For example:

```python
10 + 20
```

performs numeric addition, while:

```python
"Hello " + "Python"
```

performs string concatenation.

This is called **operator overloading**.

````python
print(10 + 20)
print("Hello " + "Python")
print([1, 2] + [3, 4])
````

### Operator Overloading in Our Own Classes

Special methods (also called **dunder methods**) allow us to define how operators behave for our objects.

For example:

```python
__add__()
```

controls the behavior of the `+` operator for a class.

````python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def display(self):
        print(f"({self.x}, {self.y})")


p1 = Point(10, 20)
p2 = Point(5, 15)

p3 = p1 + p2

p3.display()
````

### Explanation

When Python sees:

```python
p3 = p1 + p2
```

it effectively uses:

```python
p1.__add__(p2)
```

Our class defines what `+` means for two `Point` objects.

So:

```text
(10, 20) + (5, 15)
= (15, 35)
```

This is operator overloading and is a form of polymorphic behavior.

## 7. Another Operator Overloading Example

Let's create a `Book` class and define how two books can be compared based on their prices.

````python
class Book:
    def __init__(self, title, price):
        self.title = title
        self.price = price

    def __lt__(self, other):
        return self.price < other.price


book1 = Book("Python Basics", 500)
book2 = Book("Advanced Python", 800)

print(book1 < book2)
````

Here `__lt__()` defines the behavior of the `<` operator for `Book` objects.

```python
book1 < book2
```

is interpreted using:

```python
book1.__lt__(book2)
```

## 8. Method Overloading in Python

In languages such as Java or C++, method overloading usually means defining multiple methods with the same name but different parameters.

For example:

```text
add(int, int)
add(int, int, int)
```

Python does **not** support traditional method overloading by defining multiple methods with the same name.

If we define the same method multiple times in a class, the **last definition replaces the previous one**.

Instead, Python commonly achieves similar behavior using:

- Default arguments
- `*args`
- `**kwargs`


````python
class Calculator:
    def add(self, a, b=0, c=0):
        return a + b + c


calc = Calculator()

print(calc.add(10, 20))
print(calc.add(10, 20, 30))
````

### Using `*args`

Another flexible approach is to accept any number of arguments.

````python
class Calculator:
    def add(self, *numbers):
        return sum(numbers)


calc = Calculator()

print(calc.add(10, 20))
print(calc.add(10, 20, 30))
print(calc.add(1, 2, 3, 4, 5))
````

## 9. Polymorphism Through Inheritance

Let's look at a more realistic example.

Suppose we have different types of employees. Each employee has a `calculate_salary()` method, but the calculation can be different for each type.

````python
class Employee:
    def calculate_salary(self):
        return 0


class FullTimeEmployee(Employee):
    def calculate_salary(self):
        return 50000


class PartTimeEmployee(Employee):
    def calculate_salary(self):
        return 25000


class Intern(Employee):
    def calculate_salary(self):
        return 15000


employees = [
    FullTimeEmployee(),
    PartTimeEmployee(),
    Intern()
]

for employee in employees:
    print(employee.calculate_salary())
````

### What happened?

The loop uses the same statement:

```python
employee.calculate_salary()
```

But the result depends on the actual object:

```text
FullTimeEmployee → 50000
PartTimeEmployee → 25000
Intern           → 15000
```

This allows us to write one common piece of code for different employee types.

## 10. Real-World Example: Payment System

Polymorphism is especially useful when designing systems where different objects provide the same operation.

For example, different payment methods can implement `pay()` differently.

````python
class CreditCard:
    def pay(self, amount):
        print(f"Paid ₹{amount} using Credit Card")


class UPI:
    def pay(self, amount):
        print(f"Paid ₹{amount} using UPI")


class Cash:
    def pay(self, amount):
        print(f"Paid ₹{amount} using Cash")


def make_payment(payment_method, amount):
    payment_method.pay(amount)


make_payment(CreditCard(), 1000)
make_payment(UPI(), 500)
make_payment(Cash(), 300)
````

The function `make_payment()` does not care which payment method is passed.

It only requires that the object has a `pay()` method.

This is a clean example of **duck typing + polymorphism**.

## 11. Method Overriding vs Method Overloading

| Feature | Method Overriding | Method Overloading |
|---|---|---|
| Meaning | Child class changes parent method behavior | Same method handles different parameter combinations |
| Common in Python? | Yes | Not traditionally supported |
| Usually involves inheritance? | Yes | Not necessarily |
| Python approach | Define the same method in child class | Use default arguments, `*args`, `**kwargs` |
| Example | `Dog.sound()` overrides `Animal.sound()` | `add(10, 20)` and `add(10, 20, 30)` |

**Important:** Method overriding and method overloading are different concepts.

## 12. Polymorphism vs Inheritance

These concepts are related but not the same.

### Inheritance

Inheritance allows a class to acquire properties and methods from another class.

```text
Animal
  ↓
Dog
```

### Polymorphism

Polymorphism allows the same interface or operation to behave differently for different objects.

```text
Dog.sound() → Bark
Cat.sound() → Meow
```

Inheritance can be used to achieve polymorphism, but Python's duck typing allows polymorphism even **without inheritance**.

## 13. Advantages of Polymorphism

### 1. Code Reusability
The same function can work with different objects.

### 2. Flexibility
New classes can often be added without changing existing code.

### 3. Maintainability
Common operations can be handled through a single interface.

### 4. Loose Coupling
Code can depend on what an object can do rather than its exact class.

### 5. Extensibility
Large applications can support new object types more easily.

## 14. Key Points to Remember

- **Polymorphism = many forms.**
- The same interface or operation can produce different behavior.
- **Method overriding** is a major example of runtime polymorphism.
- **Duck typing** is an important Python-specific approach to polymorphism.
- Python's built-in functions such as `len()` demonstrate polymorphic behavior.
- **Operator overloading** uses dunder methods such as `__add__()` and `__lt__()`.
- Python does not support traditional compile-time method overloading by simply defining the same method multiple times.
- Default arguments, `*args`, and `**kwargs` can provide method-overloading-like flexibility.
- Polymorphism helps make programs flexible, reusable, and easier to extend.
