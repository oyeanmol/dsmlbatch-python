# `Object`

    - Abstraction
        - Hiding the data 
    - Inheritance
        - Inherit/Derive property from any other class
    - Polymorphism
        - poly > Many
        - morphism > forms
        - One thing which have many forms
    - Encapsulation
        - Binding all the data with the security into a class

```python
# Class
class Shoes:

    # Attribute, Variable
    # Class Variable
    Brand_Name = "Puma"
    Color_Name = "White"

    # init Constructor - Activates the Class
    def __init__(self, color, size):
        # Instance Variable
        self.color = color
        self.size = size
```

```python
cus1 = Shoes("Black", 8) # object 1
cus2 = Shoes("Pink", 9) # Object 2
cus3 = Shoes("Red", 7) # Object 3

print(cus2.Brand_Name) #Accessing Class Attributes
print(cus1.Brand_Name)
print(cus3.Brand_Name)
Shoes.__dict__   # Accessing full class data as a dictionary
```
