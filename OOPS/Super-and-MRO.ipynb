# Super() Method

```python
class Person:
    def show(self):
        print("Person Class")
class Hello(Person):
    def show(self):
        print("Hello Class")
class Student(Hello):
    def show(self):
        print("Student Class")
        super().show()
```

```python
a = Student()
```

```python
a.show()
```

```python
class Person:
    def __init__(self, name):
        self.name = name
        print("Person constructor")


class Student(Person):
    def __init__(self, name, roll):
        super().__init__(name)
        self.roll = roll
        print("Student constructor")
```

```python
s1 = Student("Anmol", 101)
```

```python
class Person:
    def __init__(self, name):
        self.name = name
        print("Person constructor called")
    
    def introduce(self):
        print(f"My name is {self.name}")

class Student(Person):
    def __init__(self, name, roll_no):
        super().__init__(name)
        self.roll = roll_no

    def introduce(self):
        super().introduce()
        print(f"My roll no is: {self.roll}")

class College(Student):
    def __init__(self, name, roll, college_name):
        super().__init__(name, roll)
        self.collegeName = college_name

    def introduce(self):
        super().introduce()
        print(f"My College name is: {self.collegeName}")

        
```

```python
ob1 = College("Anmol", 101, "GRRAS")
```

```python
ob1.introduce()
```
