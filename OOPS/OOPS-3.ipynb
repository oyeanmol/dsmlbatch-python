# Destructor in Python

Destructor ka purpose object ke cleanup/finalization se related hota hai

using: `__del__()`

````python
class Student:

    def __init__(self):
        print(f"{self}, Object created class initiated")

    def __del__(self):
        print(f"{self}, using del our object detroyed!")
````

````python
a = Student()
````

````python
del a
````

````python
del a
````

## del kya karta hai?
`
del s1
`
del object ko directly "destroy" karne ka guaranteed command nahi hai.

It deletes the reference s1.

If that was the last reference to the object, the object can become unreachable and Python may then finalize it, potentially calling __del__().

````python
class Student:

    def __del__(self):
        print("Destructor called")


s1 = Student()

del s1
````

Python mein __del__() ka exact execution time guaranteed nahi hota.

Python ka garbage collector / reference counting implementation object ko kab clean up karega, ye situation aur Python implementation par depend kar sakta hai.

# Class Method

A class method works with the class rather than a particular object.

It is created using:

# @classmethod  - decorator

````python
class Student:

    college = "Pradeep"

    @classmethod
    def show_college(cls):
        print(cls.college)
````

````python
Student.show_college()
````

# Example: Student Class

````python
class Student:

    school = "ABC School"   # Class Variable

    def __init__(self, name, age):
        self.name = name    # Object Variable
        self.age = age

    # Instance Method
    def show_student(self):
        print("Name:", self.name)
        print("Age:", self.age)

    # Class Method
    @classmethod
    def change_school(cls, new_school):
        cls.school = new_school


# Creating Objects
student1 = Student("Rahul", 20)
student2 = Student("Aman", 21)

# Object method
student1.show_student()
student2.show_student()

# Class Method
print("Old School:", Student.school)

Student.change_school("GRRAS")

print("New School:", Student.school)
````

````python
class Pradeep:
    p = "pradeep"

    def methodName(self):
        print(self)

    @classmethod
    def clmethod(cls, other):
        Pradeep.p = other
````

````python
obj1 = Pradeep()
obj1.methodName()
print(f"Old name: {Pradeep.p}")
Pradeep.clmethod("Sen")
print(f"New name: {Pradeep.p}")
````

# Static Method

A static method is a method that doesn't need access to either the object or the class.

It uses the @staticmethod decorator.

````python
class Calculator:

    @staticmethod
    def add(a, b):
        return a + b
````

````python
result = Calculator.add(20, 20)

print(result)
````

````python
class Math:

    @staticmethod
    def is_even(number):
        return number % 2 == 0


print(Math.is_even(10))
````

````python
class Product:

    GST_RATE = 18

    def __init__(self, name, price):
        self.name = name
        self.price = price

    # Instance Method
    def display(self):
        print("Product:", self.name)
        print("Price:", self.price)

    # Static Method
    @staticmethod
    def is_valid_price(price):
        return price > 0

    # Static Method
    @staticmethod
    def calculate_discount(price, discount_percent):
        discount = price * discount_percent / 100
        return price - discount

    # Static Method
    @staticmethod
    def calculate_gst(price, gst_rate):
        gst = price * gst_rate / 100
        return gst

    # Static Method
    @staticmethod
    def is_valid_pincode(pincode):
        return len(str(pincode)) == 6
````

````python
class GTA5:
    health_percentage = "100%"
    def __init__(self,  name):
        self.name = name

    def canWalk(self):
        print(f"{self.name} can walk")

    @classmethod
    def health(cls, shoot):
        GTA5.health_percentage = shoot

    @staticmethod
    def car():
        print(f"Sammer Can drive a car")
````

````python
print(f" Before shoting {p1.name} : {GTA5.health_percentage}")
p1.canWalk()
````
