```python
from abc import ABC, abstractmethod
class BankApplication(ABC):

    def database(self):
        print("Connected to database")

    @abstractmethod
    def security(self):
        pass
```

```python
class MobileApplication(BankApplication):

    def mobile_login(self):
        print("Login to mobile application")        

    def security(self):
        print("Mobile Security lag gyi idhar")
```

```python
mob = MobileApplication()
```

```python
mob.mobile_login()
```

```python
mob.database()
```

```python
class WebApplication(BankApplication):
    def web_login(self):
        print("Login to web application")

    def security(self):
        print("web me bhi lag gyi security")
```

```python
web = WebApplication()
```

```python
web.database()
```

```python
web.security()
```

```python
from abc import ABC, abstractmethod
class Bank(ABC):

    @abstractmethod
    def pin(self):
        pass

    @abstractmethod
    def card(self):
        pass

    @abstractmethod
    def withdraw(self):
        pass

    @abstractmethod
    def deposit(self):
        pass
```

```python
class ATM(Bank):

    acc_balance = 0

    def __init__(self):
        print("Welcome to ATM Machine")

    def card(self):
        print("Card Entered!")

    def pin(self):
        print("Pin Entered!")

    def withdraw(self):
        draw = int(input("Enter the amount you want to withdraw: "))
        if self.acc_balance < draw and self.acc_balance == 0:
            print("Insufficient Balance!")
        else:
            print("Amount Withdraw: ", draw)
        self.acc_balance = self.acc_balance - draw


    def deposit(self):
        dep = int(input("Enter the amount you want to deposit: "))
        self.acc_balance = self.acc_balance + dep
        if dep == 0:
            print("Enter proper amount")
        else:
            print(f"Amount Deposited: {dep}")

    def check_balance(self):
        print(f"Balance: {self.acc_balance}")
```

```python
cus1 = ATM()
```

```python
cus1.acc_balance
```

```python
cus1.deposit()
```

```python
cus1.check_balance()
```

```python
cus1.withdraw()
```

```python
cus1.check_balance()
```
