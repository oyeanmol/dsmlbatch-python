```python
# Multiple Inheritance with constuctors
class A:
    def __init__(self, name):
        print("A constructor")


class B:
    def __init__(self, age):
        print("B constructor")


class C(A, B):
    # MRO
    def __init__(self):
        print("C constructor")
        A.__init__(self)
        B.__init__(self)
```

```python
class A:
    def __init__(self, name):
        self.name = name
        print("A constructor")
    def method1(self):
        print("Method 1")


class B:
    def __init__(self, age):
        self.age = age
        print("B constructor")


class C(A, B):
    def __init__(self, salary, name, age):
        self.salary = salary
        # Explicit Calling the constructors
        A.__init__(self, name)
        B.__init__(self,age)
        print("C Constructor")

```

```python
ob1 = C(50000, "Anmol", 26)
```

```python
print(obj1.salary)
```

```python
class Time:

    def __init__(self, hours, minutes):
        self.hours = hours
        self.minutes = minutes

    def addTime(self, t1, t2):
        total_minutes = (t1.hours * 60 + t1.minutes) + (t2.hours * 60 + t2.minutes)
        self.hours = total_minutes // 60
        self.minutes = total_minutes % 60

    def displayTime(self):
        print(f"{self.hours} hr {self.minutes} min")

    def displayMinutes(self):
        total_minutes = self.hours * 60 + self.minutes
        print(total_minutes, " minutes")
```

```python
t1 = Time(2, 30)
t2 = Time(1, 40)

result = Time(0, 0)
result.addTime(t1, t2)
result.displayTime()
result.displayMinutes()
```

```python
class Person:
    def __init__(self, name):
        self.name = name
        print("Person constructor called")

    def introduce(self):
        print(f"My name is {self.name}")


class Student(Person):
    def __init__(self, name, roll_no):
        super().__init__(name)
        self.roll_no = roll_no
        print("Student constructor called")

    def introduce(self):
        super().introduce()
        print(f"My roll number is {self.roll_no}")


class CollegeStudent(Student):
    def __init__(self, name, roll_no, college):
        super().__init__(name, roll_no)
        self.college = college
        print("CollegeStudent constructor called")

    def introduce(self):
        super().introduce()
        print(f"My college is {self.college}")


s1 = CollegeStudent("Anmol", 101, "Poornima College")

s1.introduce()
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def __lt__(self, other):
        if self.age < other.age:
            print(f"{self.name} is lower than {other.name}")
        else:
            print(f"{self.name} is greater than {other.name}")
```

```python
p1 = Person("Anmol", 20)
p2 = Person("Rahul", 25)
```

```python
p1 < p2
```

```python
Person.__lt__(p1, p2)
```
