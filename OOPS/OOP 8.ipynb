# Access Specifiers

1. Public
2. Protected
3. Private
   

```python
class Specifiers:
    pub = "Public Variable"
    _pro = "Protected Variable"
    __private = "Private Variable"

    def show(self):
        print(f"{self.pub}")
        print(f"{self._pro}")
        print(f"{self.__private}")
```

```python
obj1 = Specifiers()
```

```python
obj1.show()
```

```python
obj1.pub
```

```python
obj1._pro
```

```python
obj1.__pri
```

```python
class New:
    __slots__ = ["name", "age"]
    def __init__(self, name, age):
        self.name = name
        self.age = age
```

```python
o = New("Anmol", 23)
```

```python
o.name
```

```python
o.age
```
