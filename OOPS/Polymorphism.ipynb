### Overriding 
Pahle mere paas ek function tha jisko maine change kardiya with the same name

Ex: greet()

greet(name)

greet(name, age)

```python
def greet():
    print("Hello World")
```

```python
def greet(name):
    print(f"Hello {name}")
```

```python
def greet(name, age):
    print(f"Hello {name}")
    print(f"Your Age {age}")
```

```python
greet("Anmol", 25)
```

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __lt__(self, other):
        if self.age < other.age:
            return True
        else:
            return False
        
```

```python
p1 = Student("Anmol", 25)
p2 = Student("Pradeep", 22)
```

```python
p1 < p2
```

```python
Student.__lt__(p1, p2)
```

```python
p1 > p2
```
