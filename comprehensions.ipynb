> **Note:** Python does not have a dedicated tuple-comprehension syntax. Parentheses around a comprehension create a **generator expression**. A tuple can be created from it using `tuple()`.


## 1. What is Comprehension?

A comprehension is a compact way to create a collection from an iterable.

### General idea

Instead of writing:

```python
result = []

for item in iterable:
    result.append(expression)
```

we can often write:

```python
result = [expression for item in iterable]
```

The same basic idea can be used with:

- `[]` → List Comprehension
- `{}` → Set Comprehension
- `{key: value}` → Dictionary Comprehension
- `tuple(...)` → Tuple from a generator expression


## 2. List Comprehension

A **list comprehension** creates a new list in a single expression.

### Syntax

```python
[expression for item in iterable]
```

### With a condition

```python
[expression for item in iterable if condition]
```

### With `if-else`

```python
[expression_if_true if condition else expression_if_false for item in iterable]
```


````python
# Basic list comprehension

numbers = [1, 2, 3, 4, 5]

squares = [num ** 2 for num in numbers]

print(squares)
````

### Equivalent normal `for` loop

The following two approaches produce the same result.

````python
numbers = [1, 2, 3, 4, 5]

squares = []

for num in numbers:
    squares.append(num ** 2)

print(squares)
````

````python
# List comprehension version

numbers = [1, 2, 3, 4, 5]

squares = [num ** 2 for num in numbers]

print(squares)
````

## 3. List Comprehension with Strings

A comprehension can iterate over strings because strings are iterable.

````python
word = "PYTHON"

letters = [char for char in word]

print(letters)
````

````python
word = "python"

uppercase = [char.upper() for char in word]

print(uppercase)
````

## 4. List Comprehension with a Range

````python
# Create numbers from 1 to 10

numbers = [num for num in range(1, 11)]

print(numbers)
````

````python
# Create even numbers from 1 to 20

even_numbers = [num for num in range(1, 21) if num % 2 == 0]

print(even_numbers)
````

````python
# Create odd numbers from 1 to 20

odd_numbers = [num for num in range(1, 21) if num % 2 != 0]

print(odd_numbers)
````

## 5. List Comprehension with `if`

Only elements satisfying the condition are included.

````python
numbers = [10, 15, 20, 25, 30, 35]

greater_than_20 = [num for num in numbers if num > 20]

print(greater_than_20)
````

````python
# Select names having more than 4 characters

names = ["Aman", "Raj", "Anmol", "Priya", "Rohit", "Neha"]

long_names = [name for name in names if len(name) > 4]

print(long_names)
````

## 6. List Comprehension with `if-else`

When both outcomes are required, use `if-else` **before** the `for`.

````python
numbers = [1, 2, 3, 4, 5, 6]

result = ["Even" if num % 2 == 0 else "Odd" for num in numbers]

print(result)
````

````python
marks = [35, 78, 42, 90, 25, 66]

result = ["Pass" if mark >= 40 else "Fail" for mark in marks]

print(result)
````

## 7. Nested List Comprehension

A nested comprehension can be used when working with nested loops or nested lists.

### Example

Create all pairs using two lists.

````python
numbers1 = [1, 2, 3]
numbers2 = [10, 20, 30]

pairs = [(x, y) for x in numbers1 for y in numbers2]

print(pairs)
````

### Flattening a nested list

````python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

flattened = [num for row in matrix for num in row]

print(flattened)
````

## 8. Set Comprehension

A **set comprehension** creates a set using a similar syntax to list comprehension.

### Syntax

```python
{expression for item in iterable}
```

Sets automatically remove duplicate values.

````python
numbers = [1, 2, 2, 3, 4, 4, 5]

unique_numbers = {num for num in numbers}

print(unique_numbers)
````

````python
# Squares using set comprehension

numbers = range(1, 6)

squares = {num ** 2 for num in numbers}

print(squares)
````

## 9. Set Comprehension with Conditions

````python
numbers = range(1, 21)

even_set = {num for num in numbers if num % 2 == 0}

print(even_set)
````

````python
words = ["python", "java", "python", "c", "java", "javascript"]

long_words = {word for word in words if len(word) > 3}

print(long_words)
````

### Important difference: List vs Set

A list preserves duplicate values, while a set stores only unique values.

````python
numbers = [1, 2, 2, 3, 3, 3]

list_result = [num for num in numbers]
set_result = {num for num in numbers}

print("List:", list_result)
print("Set :", set_result)
````

# 10. Dictionary Comprehension

A **dictionary comprehension** creates a dictionary in a compact way.

### Syntax

```python
{key_expression: value_expression for item in iterable}
```

### With condition

```python
{key_expression: value_expression for item in iterable if condition}
```


````python
# Create a dictionary of numbers and their squares

numbers = range(1, 6)

squares = {num: num ** 2 for num in numbers}

print(squares)
````

````python
# Create a dictionary from two lists

names = ["Aman", "Riya", "Anmol"]
marks = [75, 82, 90]

student_marks = {name: mark for name, mark in zip(names, marks)}

print(student_marks)
````

## 11. Dictionary Comprehension with Conditions

````python
marks = {
    "Aman": 75,
    "Riya": 35,
    "Anmol": 90,
    "Rahul": 42
}

passed_students = {
    name: mark
    for name, mark in marks.items()
    if mark >= 40
}

print(passed_students)
````

## 12. Dictionary Comprehension with `if-else`

````python
marks = {
    "Aman": 75,
    "Riya": 35,
    "Anmol": 90,
    "Rahul": 42
}

result = {
    name: "Pass" if mark >= 40 else "Fail"
    for name, mark in marks.items()
}

print(result)
````

## 13. Swapping Keys and Values

Dictionary comprehension can also be used to reverse a dictionary when values are unique.

````python
data = {
    "a": 1,
    "b": 2,
    "c": 3
}

reversed_data = {value: key for key, value in data.items()}

print(reversed_data)
````

## 14. Filtering a Dictionary

````python
prices = {
    "Laptop": 60000,
    "Mouse": 800,
    "Keyboard": 1500,
    "Monitor": 12000
}

expensive = {
    item: price
    for item, price in prices.items()
    if price > 5000
}

print(expensive)
````

# 15. Tuple Comprehension

### Important concept

Python does **not** provide a true tuple-comprehension syntax.

For example:

```python
(x for x in range(5))
```

creates a **generator expression**, not a tuple.

To create a tuple, pass the generator expression to `tuple()`:

```python
tuple(x for x in range(5))
```


````python
# Generator expression

result = (num ** 2 for num in range(1, 6))

print(result)
print(type(result))
````

````python
# Convert generator expression into a tuple

result = tuple(num ** 2 for num in range(1, 6))

print(result)
print(type(result))
````

## 16. Tuple Creation with Conditions

````python
numbers = range(1, 11)

even_tuple = tuple(
    num for num in numbers
    if num % 2 == 0
)

print(even_tuple)
````

````python
numbers = range(1, 11)

result = tuple(
    "Even" if num % 2 == 0 else "Odd"
    for num in numbers
)

print(result)
````

## 17. Why Use Generator Expressions?

A generator expression produces values lazily instead of creating the entire collection immediately.

This can be useful when working with large amounts of data.

```python
generator = (x ** 2 for x in range(1000000))
```

The values can then be consumed one at a time.

```python
for value in generator:
    print(value)
```

If you actually need a tuple, use `tuple()` around the expression.


# 18. Comparing the Four Types

| Type | Syntax | Duplicate Values | Mutable |
|---|---|---|---|
| List | `[expression for ...]` | Yes | Yes |
| Set | `{expression for ...}` | No | No |
| Dictionary | `{key: value for ...}` | Keys: No | Yes |
| Tuple | `tuple(expression for ...)` | Yes | No |

Remember:

- List comprehension → creates a list.
- Set comprehension → creates a set.
- Dictionary comprehension → creates a dictionary.
- Generator expression → creates a generator.
- `tuple(generator_expression)` → creates a tuple.


# 19. Comprehension with Functions

A function can be called inside a comprehension.

````python
def square(num):
    return num ** 2

numbers = [1, 2, 3, 4, 5]

result = [square(num) for num in numbers]

print(result)
````

````python
def is_even(num):
    return num % 2 == 0

numbers = range(1, 11)

even_numbers = [num for num in numbers if is_even(num)]

print(even_numbers)
````

# 20. String Processing with Comprehensions

````python
sentence = "Python is easy to learn"

words = sentence.split()

word_lengths = {word: len(word) for word in words}

print(word_lengths)
````

````python
sentence = "Python Programming"

vowels = [char for char in sentence.lower() if char in "aeiou"]

print(vowels)
````

# 21. Removing Spaces from a String

````python
text = "P y t h o n"

result = "".join(char for char in text if char != " ")

print(result)
````

# 22. Nested Dictionary Comprehension

Dictionary comprehensions can also contain nested loops.

````python
students = ["Aman", "Riya"]
subjects = ["Python", "SQL"]

student_subjects = {
    student: {subject: 0 for subject in subjects}
    for student in students
}

print(student_subjects)
````

# 23. Nested List Comprehension: Matrix

Create a 3 × 3 matrix.

````python
matrix = [[0 for col in range(3)] for row in range(3)]

print(matrix)
````

### Create a multiplication table using nested comprehension

````python
table = [
    [row * col for col in range(1, 6)]
    for row in range(1, 6)
]

for row in table:
    print(row)
````

# 24. Practical Example: Student Marks

Suppose we have student marks and want to perform multiple operations using comprehensions.

````python
marks = [45, 78, 32, 90, 67, 29, 88]

# Add 5 grace marks
updated_marks = [mark + 5 for mark in marks]

# Keep marks within 100
updated_marks = [min(mark + 5, 100) for mark in marks]

# Find passed marks
passed_marks = [mark for mark in updated_marks if mark >= 40]

print("Updated:", updated_marks)
print("Passed :", passed_marks)
````

# 25. Practical Example: Product Prices

````python
products = {
    "Laptop": 60000,
    "Mouse": 800,
    "Keyboard": 1500,
    "Monitor": 12000
}

# Apply 10% discount
discounted_prices = {
    product: price * 0.90
    for product, price in products.items()
}

print(discounted_prices)
````

# 26. Practical Example: Extracting Data

Suppose we have a list of dictionaries.

````python
students = [
    {"name": "Aman", "marks": 75},
    {"name": "Riya", "marks": 35},
    {"name": "Anmol", "marks": 90},
    {"name": "Rahul", "marks": 42}
]

# Extract names
names = [student["name"] for student in students]

# Extract students who passed
passed = [
    student["name"]
    for student in students
    if student["marks"] >= 40
]

print("Names:", names)
print("Passed:", passed)
````

# 27. Common Mistakes

## Mistake 1: Incorrect placement of `if-else`

Correct:

```python
result = ["Even" if x % 2 == 0 else "Odd" for x in numbers]
```

For filtering only, `if` comes after the `for`:

```python
result = [x for x in numbers if x % 2 == 0]
```

---

## Mistake 2: Confusing set and dictionary comprehension

```python
{x for x in numbers}          # Set
{x: x**2 for x in numbers}   # Dictionary
```

---

## Mistake 3: Calling a tuple comprehension

This:

```python
(x for x in numbers)
```

is a generator expression, not a tuple.

Use:

```python
tuple(x for x in numbers)
```

---

## Mistake 4: Making comprehensions unnecessarily complicated

Comprehensions should improve readability. If a comprehension becomes difficult to understand, a normal `for` loop may be better.


# 28. Comprehension vs Normal Loop

### Normal loop

```python
squares = []

for num in range(1, 6):
    squares.append(num ** 2)
```

### Comprehension

```python
squares = [num ** 2 for num in range(1, 6)]
```

Comprehension is shorter, but both approaches are valid.

Use a normal loop when:

- The logic has many steps.
- Several conditions are involved.
- Readability becomes poor.
- You need multiple statements inside the loop.

# 32. Final Summary

Comprehensions are useful for creating collections concisely.

### List Comprehension

```python
[x * 2 for x in numbers]
```

Creates a list.

### Set Comprehension

```python
{x * 2 for x in numbers}
```

Creates a set and removes duplicates.

### Dictionary Comprehension

```python
{x: x * 2 for x in numbers}
```

Creates key-value pairs.

### Tuple

```python
tuple(x * 2 for x in numbers)
```

Creates a tuple from a generator expression.

### Most important pattern

```python
[expression for item in iterable if condition]
```

And when an `if-else` is needed:

```python
[expression_if_true if condition else expression_if_false
 for item in iterable]
```

The key is to understand the normal `for` loop first and then convert simple loops into comprehensions.
