# Project: Student Attendance Management System

```python
header = "Date, Topic"
n = int(input("Enter Number of Students: "))
for i in range(n):
    name = input(f" Enter name {i + 1}: ")
    header += f", {name}"
else:
    header += "\n"

f = open("Project_Attendance.csv", 'w')
f.write(header)
f.close()

file = open("Project_Attendance.csv", 'r')
hd = file.readlines()
file.close()
names = hd[0].strip().split(', ')[2:]

main = open("Project_Attendance.csv", 'a')
date = datetime.date.today()
topic = input("Enter topic: ")
data = f"{date}, {topic}"

for i in names:
    att = input(f"Enter Attandence of {i} (P/A): ")
    data += f", {att}"

else:
    data += "\n"
    main.write(data)
    main.close()
```
