```python
students = []

# Taking names of 10 students
print("Enter names of 10 students:")

for i in range(10):
    name = input("Enter name of student " + str(i + 1) + ": ")
    students.append({"name": name})


# Taking roll number and marks
for student in students:

    # Taking valid roll number
    roll = int(input("Enter roll number of " + student["name"] + ": "))

    while roll < 1000 or roll > 9999:
        print("Invalid roll number. Please enter again.")
        roll = int(input("Enter roll number of " + student["name"] + ": "))

    student["roll"] = roll

    # Taking marks of 2 subjects
    student["marks1"] = int(input("Enter marks in Subject 1: "))
    student["marks2"] = int(input("Enter marks in Subject 2: "))


# Calculations
count = 0
highest_total = -1
highest_roll = 0

for student in students:

    if student["marks1"] >= 40 and student["marks2"] >= 40:

        total = student["marks1"] + student["marks2"]

        if total > 100:
            count = count + 1

        if total > highest_total:
            highest_total = total
            highest_roll = student["roll"]


print("Number of students whose total is more than 200:", count)
print("Roll number of student with highest total:", highest_roll)
```
