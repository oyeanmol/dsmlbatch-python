```python
def fun():
    try:
        print("Executed in try block")
        return 0
    except Exception as e:
        return 100
    else:
        print("Excecuted in else block")
        return 1000
    finally:
        print("Excuted in Finally block")
        return 10000

print(fun())
```

```python
class XYZERROR(Exception):
    pass

balance = 1000
try:
    withdrawl = int(input("Enter any amount: "))
    if withdrawl > balance:
        raise XYZERROR("Paise kam hai")
except XYZERROR as e:
    print(f"Error: {e}")
else:
    print("nikal gye paise")
finally:
    print("Take your ATM Card")
```
